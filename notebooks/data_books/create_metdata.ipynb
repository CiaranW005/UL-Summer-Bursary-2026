Test notebook for how to create metadata for the project

```python
import os
import pandas as pd

from pathlib import Path
```

```python
folder = Path("../../data/mvtec_ad")

print(folder.exists())
```

```python
images = []

# Go through each category in the dataset
for cat in folder.iterdir():
    if not cat.is_dir():
        continue

    #print(cat.name)

    # Within each category, go through each split (train/test)
    for split in cat.iterdir():
        if not split.is_dir():
            continue
        
        # Ignore ground truth for now as it's not what we're interested in
        if split.name == "ground_truth":
            continue
        
        #print(f"    {split.name}")

        # Within each split, go through each type (good/defect)
        for type in split.iterdir():    
            #print(f"        {type.name}")

            for img in type.iterdir():
                #print(f"            {img.name}")

                image_record = {
                    "path": str(img),
                    "category": cat.name,
                    "split": split.name,
                    "type": type.name,
                    "label": 0 if type.name == "good" else 1
                }

                images.append(image_record)
```

```python
for img in images:
    print(img)
```

```python
df = pd.DataFrame(images)

df
```

```python
os.makedirs("../../data", exist_ok=True)

df.to_csv("../../data/metadata.csv", index=False)
```
